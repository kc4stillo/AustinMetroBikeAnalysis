```python
import pandas as pd
import altair as alt

alt.data_transformers.disable_max_rows()

# load your dataset
df = pd.read_csv(
    "/Users/zainsyed/Downloads/metro_trips.txt",
    parse_dates=["checkout_time"],
    low_memory=False
)

# take a sample for speed
df_sample = df.sample(100000, random_state=1)

# extract checkout hour
df_sample["checkout_hour"] = df_sample["checkout_time"].dt.hour

# dropdown filter for pass type
pass_dropdown = alt.binding_select(
    options=df_sample["pass_type"].dropna().unique().tolist(),
    name="Pass Type:"
)
pass_select = alt.selection_point(fields=["pass_type"], bind=pass_dropdown)

# interactive bar chart
chart = (
    alt.Chart(df_sample)
    .mark_bar()
    .encode(
        x=alt.X("checkout_hour:O", title="Hour of Day"),
        y=alt.Y("count():Q", title="Number of Rides"),
        color=alt.Color("checkout_hour:O", legend=None, scale=alt.Scale(scheme="blues")),
        tooltip=["checkout_hour", "count()"]
    )
    .add_params(pass_select)
    .transform_filter(pass_select)
    .properties(title="Hourly Ridership by Pass Type", width=700, height=400)
    .interactive()
)

chart
```

```python
# add weekday column
df_sample["weekday"] = df_sample["checkout_time"].dt.day_name()

# create weekday-hour heatmap
heatmap = (
    alt.Chart(df_sample)
    .mark_rect()
    .encode(
        x=alt.X("checkout_hour:O", title="Hour of Day"),
        y=alt.Y("weekday:N", sort=["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"]),
        color=alt.Color("count():Q", title="Number of Rides", scale=alt.Scale(scheme="blues")),
        tooltip=["weekday", "checkout_hour", "count()"]
    )
    .properties(title="Ridership by Hour and Weekday", width=600, height=300)
)

heatmap.interactive()
```
